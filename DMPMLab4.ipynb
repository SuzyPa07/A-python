```python
import numpy as np
import pandas as pd

# Sample data
data = np.array([10, 20, 30, 40, 50, 100, 150, 200])

# Min-Max Normalization
min_val = np.min(data)
max_val = np.max(data)
min_max_normalized_data = (data - min_val) / (max_val - min_val)

print("Original Data:", data)
print("Min-Max Normalized Data:", min_max_normalized_data)
```

```python
# Z-score Normalization
mean_val = np.mean(data)
std_val = np.std(data)
z_score_normalized_data = (data - mean_val) / std_val

print("Original Data:", data)
print("Z-score Normalized Data:", z_score_normalized_data)
```

```python
# Decimal Scaling Normalization
max_abs_val = np.max(np.abs(data))
j = np.ceil(np.log10(max_abs_val))
decimal_scaled_data = data / (10**j)

print("Original Data:", data)
print("Decimal Scaled Data:", decimal_scaled_data)
```

COVARIANCE

```python
import numpy as np

X = data
Y = np.array([5, 12, 18, 25, 30, 55, 70, 95]) + np.random.normal(0, 2, len(data))

print("Data (X):", X)
print("Data (Y):", Y)

# Calculate covariance
covariance_xy = np.cov(X, Y)[0, 1]

print(f"\nCovariance between X and Y: {covariance_xy:.2f}")
```

CORELATION

```python
# Calculate Pearson correlation coefficient
correlation_coefficient_xy = np.corrcoef(X, Y)[0, 1]

print("Data (X):", X)
print("Data (Y):", Y)
print(f"\nPearson Correlation Coefficient between X and Y: {correlation_coefficient_xy:.2f}")
```
