```python
import numpy as np
import pandas as pd
from statistics import mean, median, mode, variance, stdev
data=[20,40,60,80]
print("Mean:",mean(data))
print("Median:",median(data))
print("Mode:",mode(data))
print("Variance:",variance(data))
print("Standard deviation:",stdev(data))
```

```python
import math
numbers = list(map(int, input("Enter numbers: ").split()))
n = len(numbers)

# Mean
total = 0
for num in numbers:
    total += num
mean = total / n

# Median
numbers.sort()
if n % 2 == 0:
    median = (numbers[n//2 - 1] + numbers[n//2]) / 2
else:
    median = numbers[n//2]

# Mode
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

max_count = max(frequency.values())
modes = []

for key, value in frequency.items():
    if value == max_count:
        modes.append(key)

# Variance
variance = 0
for num in numbers:
    variance += (num - mean) ** 2
variance = variance / n

# Standard Deviation
std = math.sqrt(variance)

# Output
print("\nStatistical Measures")
print("--------------------")
print("Mean =", mean)
print("Median =", median)

if len(modes) == len(frequency):
    print("Mode = No mode")
else:
    print("Mode =", modes)

print("Variance =", variance)
print("Standard Deviation =", std)
```

```python
data=[20,40,60,80]
n=len(data)

#Mean
mean=sum(data)/n
print("Mean:",mean)

#Median
data.sort()
if n%2==0:
  median=(data[n//2-1]+data[n//2])/2
else:
  median=data[n//2]
print("Median:",median)

#Mode
count={}
for i in data:
  count[i]=count.get(i,0)+1
  mode=max(count,key=count.get)
print("Mode:",mode)

#Variance
s=0
for i in data:
  s+=(i-mean)**2
variance=s/n
print("Variance:",variance)

#Standard Deviation
std=variance**0.5
print("Standard Deviation:",std)

```
